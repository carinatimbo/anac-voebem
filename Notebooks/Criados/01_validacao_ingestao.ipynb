## 🔍 Análise de Qualidade de Dados - sigla_iata

```python
%sql
-- Análise da distribuição de valores na coluna sigla_iata
SELECT 
  sigla_iata,
  COUNT(*) AS quantidade,
  COUNT(*) * 100.0 / SUM(COUNT(*)) OVER () AS percentual
FROM voebem.bronze.empresas_nacionais
GROUP BY sigla_iata
ORDER BY quantidade DESC
LIMIT 20
```

```python
%sql
-- Exemplos de registros com sigla_iata = '..'
SELECT 
  icao,
  sigla_iata,
  razao_social,
  servico,
  situacao
FROM voebem.bronze.empresas_nacionais
WHERE sigla_iata = '..'
LIMIT 15
```

```python
%sql
-- Comparação: distribuição na tabela de empresas estrangeiras
SELECT 
  sigla_iata,
  COUNT(*) AS quantidade,
  COUNT(*) * 100.0 / SUM(COUNT(*)) OVER () AS percentual
FROM voebem.bronze.empresas_estrangeiras
GROUP BY sigla_iata
ORDER BY quantidade DESC
LIMIT 20
```

```python
%sql
-- Empresas nacionais que possuem código IATA válido (não '..')
SELECT 
  icao,
  sigla_iata,
  razao_social,
  servico,
  situacao
FROM voebem.bronze.empresas_nacionais
WHERE sigla_iata <> '..' 
  AND sigla_iata IS NOT NULL
ORDER BY razao_social
LIMIT 20
```
